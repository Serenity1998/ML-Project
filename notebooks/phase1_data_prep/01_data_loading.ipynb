# Phase 1.1: Data Loading and Sampling

Load Yelp dataset and create a manageable sample for Phase 1 development.

```python
import sys
sys.path.insert(0, '../../')

import pandas as pd
import numpy as np
from config import *
from src.data.loader import load_businesses, load_reviews, load_users
from src.data.normalizer import normalize_city_names

print(f"Loading Yelp data from {RAW_DATA_DIR}...")
```

```python
# Load sample (10% for Phase 1)
businesses = load_businesses(BUSINESS_FILE, sample=True, sample_fraction=PHASE1_SAMPLE_FRACTION)
reviews = load_reviews(REVIEW_FILE, sample=True, sample_fraction=PHASE1_SAMPLE_FRACTION)
users = load_users(USER_FILE, sample=True, sample_fraction=PHASE1_SAMPLE_FRACTION)

print(f"Loaded:")
print(f"  Businesses: {len(businesses):,}")
print(f"  Reviews: {len(reviews):,}")
print(f"  Users: {len(users):,}")
```

```python
# Normalize city names
businesses = normalize_city_names(businesses, city_column='city')
businesses.head()
```

```python
# Save sampled data
from src.data.cache import DataCache
cache = DataCache(str(CACHE_DIR))

cache.save_dataframe(businesses, 'businesses_sample')
cache.save_dataframe(reviews, 'reviews_sample')
cache.save_dataframe(users, 'users_sample')

print("✓ Sample data cached")
```
