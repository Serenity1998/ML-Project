# Phase 1.4: Feature Engineering

Build initial user and destination feature vectors.

```python
import sys
sys.path.insert(0, '../../')

import pandas as pd
from config import *
from src.data.cache import DataCache
from src.features.user_features import UserProfileBuilder
from src.features.destination_features import DestinationProfileBuilder

cache = DataCache(str(CACHE_DIR))
businesses = cache.load_dataframe('businesses_major_cities')
reviews = cache.load_dataframe('reviews_cross_city')
```

```python
# Build user profiles
user_builder = UserProfileBuilder(reviews, businesses)
sample_users = reviews['user_id'].unique()[:5]  # Sample for inspection

for user_id in sample_users:
    profile = user_builder.build_profile(user_id)
    print(f"User {user_id}:")
    print(f"  Primary city: {profile['primary_city']}")
    print(f"  Cities visited: {len(profile['cities_visited'])}")
    print(f"  Avg rating: {profile['rating_behavior']['mean_rating']:.2f}")
    print()
```

```python
# Build destination profiles
dest_builder = DestinationProfileBuilder(businesses, reviews)
sample_cities = businesses['city'].unique()[:3]  # Sample for inspection

for city in sample_cities:
    profile = dest_builder.build_profile(city)
    print(f"City {city}:")
    if profile['stats']:
        print(f"  Businesses: {profile['stats']['num_businesses']}")
        print(f"  Avg rating: {profile['stats']['avg_rating']:.2f}")
    print()
```
