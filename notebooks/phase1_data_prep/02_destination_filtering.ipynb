# Phase 1.2: Destination Filtering

Identify cities with enough businesses and reviews to be valid recommendation destinations.

```python
import sys
sys.path.insert(0, '../../')

import pandas as pd
from config import *
from src.data.cache import DataCache
from src.data.normalizer import get_major_cities, filter_to_major_cities

cache = DataCache(str(CACHE_DIR))
businesses = cache.load_dataframe('businesses_sample')
reviews = cache.load_dataframe('reviews_sample')
```

```python
# Identify major cities
major_cities = get_major_cities(
    businesses,
    min_businesses=PHASE1_MIN_BUSINESSES_PER_CITY,
    min_reviews=PHASE1_MIN_REVIEWS_PER_CITY
)

print(f"Found {len(major_cities)} major cities:")
for city in sorted(major_cities):
    print(f"  - {city}")
```

```python
# Filter to major cities only
businesses_filtered = filter_to_major_cities(businesses, 'city', major_cities)
reviews_filtered = reviews[reviews['business_id'].isin(businesses_filtered['business_id'])]

print(f"After filtering to major cities:")
print(f"  Businesses: {len(businesses_filtered):,}")
print(f"  Reviews: {len(reviews_filtered):,}")
```

```python
# Save filtered data
cache.save_dataframe(businesses_filtered, 'businesses_major_cities')
cache.save_dataframe(reviews_filtered, 'reviews_major_cities')
cache.save_metadata({'major_cities': major_cities}, 'major_cities')

print("✓ Filtered data cached")
```
